Función mostrar playlist

```python
# Crear función mostrar playlist
def mostrar_playlist(playlist:list):
  '''
  Función -> mostrar_playlist: Imprime las canciones(diccionario) de una playlist (lista) en forma de cadena de texto.
  Param   -> playlist: Lista que contiene tres diccionarios, cada uno es un contacto con Nombre, teléfono y correo
  Return  -> No regresa nada, por lo que es None.
  '''
  numero_cancion = 0
  cancion = ""
  for cancion in playlist:
    numero_cancion += 1
    print(f'Canción # {numero_cancion}: {cancion}')
```

Función para añadir una canción a la playlist

```python
# Crear función añadir canción
def agregar_cancion(playlist:list):
  '''
  Función -> agregar_cancion: Agrega una canción nuevo (diccionario) al final de la lista de canciones
  Param   -> playlist: Lista que contiene diccionarios, cada uno es una canción con Nombre, Artista, Duración, Álbum y Año
  Return  -> None
  '''
  minutos = random.randint(1,10)
  segundos = random.randint(0, 59)
  cancion=playlist.append({'Nombre': input('Ingresa el nombre de la canción: '),
                           "Artista" : input('Ingresa el nombre del artista: '), "Duración" : f"{minutos}:{segundos:02d}",
                           "Álbum" : input('Ingresa el nombre del álbum: '),
                           "Año" : input('Ingresa el año de lanzamiento: ')})
  print("¡Canción añadida con éxito!")
```

Función para consultar canción:

```python
# Crear función consultar canción
def consultar_cancion(playlist):
  '''
  Función -> consultar_canción: Se le pide al usuario ingresar un nombre de la canción a consultar, el cuál se guarda en una variable,
  mediante un ciclo for se consulta cada canción (diccionario) de una Playlist (lista), y con un condicional if se verifica si el nombre buscado
  está "asignado" a la clave nombre, si está se dan los datos de la canción, sino, se devuelve un mensaje de Canción no encontrada
  Param   -> playlist: Lista que contiene diccionarios, cada uno es una canción con Nombre, Artista, Duración, Álbum y Año
  Return -> None
  '''
  nombre_buscado = input("Ingresa el nombre de la canción a consultar: ")
  for cancion in playlist:
    if nombre_buscado == cancion['Nombre']:
      print(f' Datos: {cancion}')
      break
  else:
    print("Canción no encontrada")
```

Función para actualizar una canción:

```python
# Crear función menú para actualizar canción
def menu_actualizar_cancion():
  '''
  Función -> menu_actualizar_cancion: Guarda en una variable una cadena de texto que contiene el menú de opciones
  Return  -> Regresa la variable donde se guardó la cadena de texto que contiene el menú de opciones
  '''
  menu_actualizar_datos_cancion = '''
  ********** Menu actualizar contacto **********
1. Actualizar Nombre
2. Actualizar Artista
3. Actualizar Álbum
4. Salir del menu
  '''
  return menu_actualizar_datos_cancion
```

Función para buscar el nombre de la canción en la playlist

```python
# Crear función buscador contacto para actualizar contacto
def buscar_cancion(playlist):
  '''
  Función -> buscar_cancion: Busca el nombre de la canción ingresada por el usuario, esto mediante un ciclo for y
  un if que compara si el nombre es igual a la clave nombre del diccionario (la canción)
  Param   -> playlist: Lista que contiene diccionarios, cada uno es una canción con Nombre, Artista, Duración, Álbum y Año
  Return  -> Regresa los datos del contacto sobre el que actualmente se está iterando
  '''
  nombre_buscado = input("Ingresa el nombre de la canción a consultar: ")
  for cancion in playlist:
    if nombre_buscado == cancion['Nombre']:
      return cancion
```

Función para actualizar el nombre la canción:

```python
# Crear función actualizar nombre para actualizar canción
def actualizar_nombre(playlist):
  '''
  Función -> actualizar_nombre: Esta función llama a la función buscar_cancion para verificar que el nombre que se quiera
  actualizar exista dentro de la playlist, si es así, se entra en la condición if que pide el nuevo nombre y se guarda en el
  diccionario para la clave Nombre, de lo contrario se imprime el mensaje que no se encontró la canción
  Param   -> playlist: Lista que contiene diccionarios, cada uno es una canción con Nombre, Artista, Duración, Álbum y Año
  Return  -> No regresa nada, por lo que es None.
  '''
  cancion = buscar_cancion(playlist)
  if cancion != None:
    clave_nombre = input('Ingresa un nuevo nombre: ')
    cancion['Nombre'] = clave_nombre
    print("Nombre actualizado:", cancion)
  else:
    print("No se encontró alguna canción con ese nombre")
```

Función para actualizar el nombre del artista:

```python
# Crear función actualizar artista para actualizar canción
def actualizar_artista(playlist):
  '''
  Función -> actualizar_artista: Esta función llama a la función buscar_cancion para verificar que el nombre que
  se quiera actualizar exista dentro de la playlisy, si es así, se entra en la condición if que pide el nuevo
  artista y se guarda en el diccionario para la clave Artista, de lo contrario se imprime el mensaje que
  no se encontró la canción
  Param   -> playlist: Lista que contiene diccionarios, cada uno es una canción con Nombre, Artista, Duración, Álbum y Año
  Return  -> No regresa nada, por lo que es None.
  '''
  cancion = buscar_cancion(playlist)
  if cancion != None:
    clave_artista = input('Ingresa un nuevo artista: ')
    cancion['Artista'] = clave_artista
    print("Artista actualizado:", cancion)
  else:
    print("No se encontró alguna canción con ese nombre")
```

Función para actualizar el Álbum

```python
# Crear función actualizar Álbum para actualizar canción
def actualizar_album(playlist):
  '''
  Función -> actualizar_album: Esta función llama a la función buscar_cancion para verificar que el nombre que
  se quiera actualizar exista dentro de la playlisy, si es así, se entra en la condición if que pide el nuevo
  álbum y se guarda en el diccionario para la clave Álbum, de lo contrario se imprime el mensaje que
  no se encontró la canción
  Param   -> playlist: Lista que contiene diccionarios, cada uno es una canción con Nombre, Artista, Duración, Álbum y Año
  Return  -> No regresa nada, por lo que es None.
  '''
  cancion = buscar_cancion(playlist)
  if cancion != None:
    clave_album = input('Ingresa un nuevo Álbum: ')
    cancion['Álbum'] = clave_album
    print("Álbum actualizado:", cancion)
  else:
    print("No se encontró alguna canción con ese nombre")
```

Función que integra las funciones para actualizar los datos de la canción:

```python
# Crear función actualizar canción
def actualizar_cancion(playlist):
  '''
  Función -> actualizar_cancion: Esta función imprime la función menu_actualizar_cancion, luego se le pide al usuario
  que ingrese una opción según lo que se vió en pantalla, con if y elif se compara que número eligió el usuario y se
  ejecuta cierta función según la opción, si la opción no coincide con algún if o elif, se imprime el mensaje: Opción no válida
  Param   -> playlist: Lista que contiene diccionarios, cada uno es una canción con Nombre, Artista, Duración, Álbum y Año
  Return  -> None
  '''
  while True:
    print(menu_actualizar_cancion())
    opcion_menu_actu_cancion = int(input("Ingrese una opción: "))

    if opcion_menu_actu_cancion == 1:
      actualizar_nombre(playlist)

    elif opcion_menu_actu_cancion == 2:
      actualizar_artista(playlist)

    elif opcion_menu_actu_cancion == 3:
      actualizar_album(playlist)

    elif opcion_menu_actu_cancion == 4:
      print("Saliendo del menú...")
      break

    else:
      print("Opción no válida")
```

Función para eliminar contacto:

```python
# Crear función eliminar cancion
def eliminar_cancion(playlist):
  '''
  Función -> eliminar_cancnion: Esta función pide el nombre a buscar y lo guarda en una variable, también se define el número
  de cancion como igual a cero, se itera sobre cada diccionario en la lista (playlist) y si el nombre buscado coincide con la
  clave Nombre del diccionario actual, se elimina todo el diccionario y se imprime el mensaje de canción eliminado. La variable
  num_cancion se va aumentando por cada iteración para utilizar la función pop. Si no se encuentra el nombre
  se imprime el mensaje: Canción no encontrada
  Param   -> playlist: Lista que contiene diccionarios, cada uno es una canción con Nombre, Artista, Duración, Álbum y Año
  Return  -> None.
  '''
  nombre_buscado_eliminar = input("Ingresa el nombre de la canción a eliminar: ")
  num_cancion = 0
  for cancion in playlist:
    if nombre_buscado_eliminar == cancion['Nombre']:
      playlist.pop(num_cancion)
      print("Canción eliminada con éxito")
      break
    num_cancion += 1
  else:
    print("Canción no encontrada")
```
